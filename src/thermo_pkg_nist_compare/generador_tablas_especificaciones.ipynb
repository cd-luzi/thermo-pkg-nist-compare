La función recibe:

- el arreglo de puntos de $x$ (obligatorio)
- el arreglo de puntos de $y$ (obligatorio)
- el valor donde se quieren comenzar las evaluaciones (opcional)
- el valor donde se quieren finalizar las evaluaciones, incluido (opcional)
- el paso con el que se evaluan los datos (opcional)
- la cantidad de decimales con las que se quieren los resultados (opcional)
- el nombre de la variable $x$ (opcional)
- el nombre de la variable $y$ (opcional)
- la unidad de la variable $x$ (opcional)
- la unidad de la variable $y$ (opcional)

Realiza una tabla para los valores entre el comienzo y el fin especificados, con el espaciado pedido y la cantidad de decimales solicitada. En caso de no especificar comienzo y fin, se utilizan el primer y último valor del arreglo de datos de entrada, x. Si no se especifica un espaciado, se evalua la función en 20 valores equiespaciados.
