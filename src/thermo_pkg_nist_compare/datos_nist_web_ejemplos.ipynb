## Ejemplo 1: Lectura web de datos del agua (sustancia y unidades por defecto) en condiciones de una sola fase.

Resumen de parámetros requeridos
* Tipo de búsqueda (parámetro: `tipo_busq`): isobárica
* Valor de presión (parámetro: `iso_value`): 0.1 MPa
* Temperatura mínima (parámetro: `rg_min`): 373 K
* Temperatura máxima (parámetro: `rg_max`): 573 K
* Delta Temperatura (parámetro: `rg_interval`): 20 K

```python
data_table = busqueda_web_nist(
    tipo_busq="IsoBar",
    iso_value=0.1,
    rg_min=373,
    rg_max=573,
    rg_interval=20,
)

print("Impresión de la tabla de prueba completa:\n")
print(data_table)
```

**Con la misma tabla, probamos algunas alternativas de selección de columnas (e impresión)**

```python
print("filas 2-5 de temperatura, usando uno de los alias: ")
for val in data_table["Temperature"][2:6]:
    print(
        f" * T = {val} {data_table.unidades[data_table.alias["Temperature"]]}"
        )

print(
    "\nImpresión de un extracto de la tabla con un subconjunto de\n"
    "propiedades desde los nombres cortos (los visibles):"
)
table_red = data_table[["T", "P", "u", "h", "s"]]
print(table_red)
```

**Si pedimos una sola columna, nos devuelve un array de numpy**

```python
h = table_red["h"]
if isinstance(h,np.ndarray):
    print("'h' es un ndarray de numpy")
else:
    print("'h' NO es un ndarray de numpy")

print(h)
```

## Ejemplo 2: Misma lectura web de datos del agua, pero cambiando las unidades (una sola fase).

Además, ingresamos dos cosas mal en el diccionario de unidades para verificar que las omita:

* `"Tunidades": "ºC"` -> x (clave errónea, debería ser `"TUnit"`)
* `"DUnit": "mol/m^3"` -> x (unidad errónea, debería ser `"mol/m3"`)

Resumen de datos requeridos
* Tipo de búsqueda (parámetro: `tipo_busq`): isobárica
* Valor de presión (parámetro: `iso_value`): 1 bar
* Temperatura mínima (parámetro: `rg_min`): 373 K
* Temperatura máxima (parámetro: `rg_max`): 573 K
* Delta Temperatura (parámetro: `rg_interval`): 20 K

```python
data_table = busqueda_web_nist(
    tipo_busq="IsoBar",
    iso_value=1,
    rg_min=373,
    rg_max=573,
    rg_interval=20,
    unidades = {
        "PUnit": "bar",
        "HUnit": "kJ/kg",
        "Tunidades": "ºC",
        "DUnit": "mol/m^3",
        },
)

print("Impresión de la tabla de prueba completa:\n")
print(data_table)
```

## Ejemplo 3: Seguimos con el agua, pero probamos los datos de equilibrio.

En estos casos la función devuelve dos tablas de tipo `PropData`

Resumen de datos requeridos
* Tipo de búsqueda (parámetro: `tipo_busq`): Equilibrio, por temperatura (presión a determinar)
* Valor de presión (parámetro: `iso_value`): No requerido

* Temperatura máxima (parámetro: `rg_max`): 473 K
* Delta Temperatura (parámetro: `rg_interval`): 10 K


```python
tabla_liq, tabla_vap = busqueda_web_nist(
    tipo_busq="SatP",
    iso_value=None,
    rg_min=293,
    rg_max=473,
    rg_interval=10,
    unidades = {
        "PUnit": "bar",
        "HUnit": "kJ/kg",
        "DUnit": "mol/m3",
        },
)

print("Propiedades de la fase líquida:\n")
print(tabla_liq)

print("\nPropiedades de la fase vapor:\n")
print(tabla_vap)
```

## Ejemplo 4: Otra sustancia

Empezamos por listar todas las sustancias disponibles con `print(Substance)`

Resumen de datos requeridos
* Tipo de búsqueda (parámetro: `tipo_busq`): isotérmica
* Valor de temperatura (parámetro: `iso_value`): 383 K
* Presión mínima (parámetro: `rg_min`): 0.5 bar
* Presión máxima (parámetro: `rg_max`): 15 bar
* Delta Presión (parámetro: `rg_interval`): 0.5 bar

```python
print(Substance)

sust = Substance.from_alias("Benceno")

data_table = busqueda_web_nist(
    tipo_busq="IsoTherm",
    iso_value=383,
    rg_min=0.5,
    rg_max=15,
    rg_interval=0.5,
    unidades = {
        "PUnit": "bar",
        "HUnit": "kJ/kg",
        },
    sustancia = sust,
)

print("\n\n________________________________")
print("Propiedades del Benceno a 383 K:\n")
print(data_table)
```
