La función recibe:

- el arreglo de puntos de $x$ (obligatorio)
- el arreglo de puntos de $y$ (obligatorio)
- el nombre de la variable $x$ (opcional)
- el nombre de la variable $y$ (opcional)
- la unidad de la variable $x$ (opcional)
- la unidad de la variable $y$ (opcional)
- el color (opcional, el color por defecto es magenta)
- el grosor de línea (opcional)
- el tamaño del marcador (opcional)

La función realiza la interpolación de los datos utilizando la función "interpolacion", utilizando un paso entre puntos de 0.05, para todo el rango de $x$ del arreglo de datos. Se grafican junto con la crurva de los valores interpolados, tambien los datos de entrada, con los cuales se realizó la interpolación, representados como hexagonos.
