La función recibe:

- el arreglo de puntos de $x$ (obligatorio),
- el arreglo de puntos de $y$ (obligatorio),
- el valor donde comienzan las evaluaciones (opcional),
- el valor donde finalizan las evaluaciones, incluido (opcional),
- el paso con el que se evalúan los datos (opcional).

La función devuelve dos arrays de NumPy: uno con los valores de $x$ y otro con los valores de $y$ interpolados.

En caso de no especificar el comienzo y el fin, se utilizan el primer y el último valor de $x$, respectivamente. Si no se especifica el paso, se usa

$$
\text{paso} = \frac{\text{fin} - \text{comienzo}}{5}
$$
