La función recibe:

- el arreglo de puntos de $x$ (obligatorio)
- el arreglo de puntos de $y$ (obligatorio)
- el valor donde se quieren comenzar las evaluaciones (opcional)
- el valor donde se quieren finalizar las evaluaciones, incluido (opcional)
- el paso con el que se evalúan los datos (opcional)

La función interpola los datos con un spline cuadrático, es decir, una función formada por polinomios de grado 2 pegados entre sí que pasa por todos los puntos $(x, y)$. Para armarlo hacen falta al menos 3 puntos. Después evalúa el spline desde el comienzo, avanzando de a un paso, sin pasarse del fin pero incluyéndolo si cae justo en un paso.

Si no se especifican el comienzo y el fin, se usan el primer y el último valor de $x$, respectivamente. Si no se especifica el paso, se usa la quinta parte del rango, es decir, se evalúa en 6 puntos equiespaciados:

$$
\text{paso} = \frac{\text{fin} - \text{comienzo}}{5}
$$

Si el comienzo y el fin coinciden, se evalúa solo en ese punto.

La función devuelve dos arrays de NumPy: uno con los puntos de $x$ donde se evaluó y otro con los valores de $y$ interpolados en esos puntos.

Hay que tener en cuenta que, fuera del rango de los datos de $x$, el spline extrapola, así que esos valores pueden no ser confiables.
