La función recibe:

- el arreglo de puntos de $x$ del set principal (obligatorio)
- el arreglo de puntos de $y$ del set principal (obligatorio)
- el subtítulo del set principal (opcional, por defecto "NIST")
- el arreglo de puntos de $x$ del set de comparación (opcional)
- el arreglo de puntos de $y$ del set de comparación (opcional)
- el subtítulo del set de comparación (opcional, por defecto "Otro")
- el valor donde se quieren comenzar las evaluaciones (opcional)
- el valor donde se quieren finalizar las evaluaciones, incluido (opcional)
- el paso con el que se evalúan los datos (opcional)
- la cantidad de decimales con las que se quieren los resultados (opcional, por defecto 3)
- el nombre de la variable $x$ (opcional)
- el nombre de la variable $y$ (opcional)
- la unidad de la variable $x$ (opcional)
- la unidad de la variable $y$ (opcional)
- si se quieren interpolar los datos o no (opcional, por defecto sí)

Si se pide interpolar, la función interpola los datos con la función `interpolacion`, que usa un spline cuadrático. Arma una tabla con los valores entre el comienzo y el fin especificados, con el espaciado pedido y la cantidad de decimales solicitada. Si no se especifican comienzo y fin, se usan el primer y el último valor del arreglo $x$. Si no se especifica un espaciado, se usa la vigésima parte del rango de $x$, es decir, $|x_{final} - x_{inicial}| / 20$. Si no se pide interpolar, se tabulan los datos tal cual y se ignoran el comienzo, el fin y el espaciado.

Si se pasa un segundo set de datos, se agrega a la derecha del primero y cada set lleva arriba su subtítulo, para identificar de dónde viene cada uno. Si los sets tienen distinto largo, las celdas que sobran quedan vacías. Si no se pasa un segundo set, o si sus arreglos $x$ e $y$ tienen distinto largo entre sí, la tabla se hace solo con el set principal y sin subtítulos.
