Pruebo extraer los datos para el agua, utilizando CoolProp. 

Para CoolProp se puede elegir trabajar con IAPWS-95 ("Water") o IAPWS-97 ("IF97::Water")

Importo la clase propdata.py para darle el mismo formato 

Si funciona, extraigo lo necesario para armar un script.py para importar al main

```python
import CoolProp.CoolProp as CP
from thermo_pkg_nist_compare import PropData
import numpy as np
```

```python
# IAPWS95 ("Water")

P = np.linspace(1e4,1e5,10)/1000    # kPa

T = np.linspace(300,400,10)    # K

u = CP.PropsSI("Umass", "P", P*1000, "T", T, "Water")/1000 # kJ/(kg K)
h = CP.PropsSI("Hmass", "P", P*1000, "T", T, "Water")/1000  # kJ/kg
s = CP.PropsSI("Smass", "P", P*1000, "T", T, "Water")/1000 # kJ/kg
    
encabezados = ["T", "P", "u", "h", "s"]

unidades = {"T": "K", "P": "kPa", "u": "kJ/kg", "h" : "kJ/kg", "s":"kJ/kg*K"}

tabla_array = np.column_stack((T, P, u, h, s))

tabla = PropData(datos=tabla_array, encabezados=encabezados, unidades=unidades)

print(tabla)
```
