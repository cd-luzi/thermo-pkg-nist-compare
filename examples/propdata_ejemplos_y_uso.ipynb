## Prueba sobre el uso de las tablas generadas con PropData cuando no proviene de la búsqueda web del NIST

Probamos dos alternativas, con datos cualesquiera, pero que sean consistentes con lo que se esperaría para el caso de propiedades termodinámicas.

### Caso 1: desde un array de numpy

```python
rng = np.random.default_rng()

tabla_array = 100*rng.random((10, 4))
encabezados = ["T", "P", "u", "s"]
unidades = {"T": "K", "P": "kPa", "u": "kJ/kg", "s":"kJ/kg*K"}

tabla = PropData(datos=tabla_array, encabezados=encabezados, unidades=unidades)

print(tabla)
```

### Caso 2: desde una lista de listas

```python
tabla_lista = [
    list(range(100, 115)),
    list(range(-5, 10)),
    list(range(250, 265)),
    list(range(360, 375)),
]

encabezados = ["T", "P", "u", "s"]
unidades = {"T": "K", "P": "kPa", "u": "kJ/kg", "s":"kJ/kg*K"}

tabla = PropData(datos=tabla_lista, encabezados=encabezados, unidades=unidades)

print(tabla)
```
