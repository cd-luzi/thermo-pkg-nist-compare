La función recibe:

- el arreglo de puntos de $x$ del set principal (obligatorio)
- el arreglo de puntos de $y$ del set principal (obligatorio)
- el nombre del set principal para la leyenda (opcional, por defecto "NIST")
- el arreglo de puntos de $x$ del set de comparación (opcional)
- el arreglo de puntos de $y$ del set de comparación (opcional)
- el nombre del set de comparación para la leyenda (opcional)
- el nombre de la variable $x$ (opcional)
- el nombre de la variable $y$ (opcional)
- la unidad de la variable $x$ (opcional)
- la unidad de la variable $y$ (opcional)
- si se grafican o no las curvas interpoladas (opcional, por defecto sí)
- el color del set principal (opcional, el color por defecto es magenta)
- el grosor de línea de las curvas (opcional)
- el tamaño del marcador del set principal (opcional)

Los datos del set principal se grafican como hexágonos del color elegido. Si se pasa un segundo set de datos, se grafica en el mismo gráfico como círculos azules, de la mitad del tamaño que los hexágonos, y se agrega una leyenda que indica el nombre de cada set. Si se grafica un solo set, no se muestra leyenda. Si no se le da nombre al set de comparación, ese set no aparece en la leyenda.

Por defecto, la función interpola cada set de datos con la función `interpolacion`, utilizando un paso entre puntos de 0.05 para todo el rango de $x$ de ese set, y grafica la curva de los valores interpolados junto con los datos de entrada. La curva del set principal es una línea continua del mismo color que sus puntos y la del set de comparación es una línea azul punteada. Si se indica que no se interpole, se grafican solamente los puntos.

Se crea, si no existe ya, una carpeta llamada "Graficos", donde se guardan todos los gráficos realizados con ésta función, en cuyo nombre se indican las variables graficadas, junto con el rango de graficación del set principal. En caso de no especificar los nombres de las variables y las unidades de las mismas, el gráfico no se guarda y se envía un mensaje de aviso para agregarlos, pero se realiza el gráfico igualmente para su visualización.
